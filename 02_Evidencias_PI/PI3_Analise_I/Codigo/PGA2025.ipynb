```python
# Armazenar texto e tabelas
textos_extraidos = []
tabelas_extraidas = []

# Abrir o PDF e extrair os dados
with pdfplumber.open("PGA2025.pdf") as pdf:
    for i, page in enumerate(pdf.pages):
        print(f"\n===== Página {i+1} =====\n")

        text = page.extract_text()
        if text:
            textos_extraidos.append(text)
            print(text[:1000])  # Exibe parte do texto
        else:
            print("Nenhum texto detectado nesta página.")

        tables = page.extract_tables()
        for table in tables:
            tabelas_extraidas.append(table)
            print("\nTabela(s) encontrada(s):")
            for row in table:
                print(row)
```

```python
# Exportar o texto extraído para um arquivo
with open("texto_extraido.txt", "w", encoding="utf-8") as f:
    for texto in textos_extraidos:
        f.write(texto + "\n\n")
print("Arquivo de texto exportado com sucesso!")
```

```python
# Exportar a primeira tabela para CSV
if tabelas_extraidas:
    with open("tabela_exemplo.csv", "w", newline='', encoding='utf-8') as f:
        writer = csv.writer(f)
        for row in tabelas_extraidas[0]:
            writer.writerow(row)
    print("Tabela exportada para 'tabela_exemplo.csv'.")
else:
    print("Nenhuma tabela encontrada para exportar.")
```

```python
# Visualizar tabelas como DataFrame
for i, tabela in enumerate(tabelas_extraidas):
    print(f"\nTabela {i+1}")
    df = pd.DataFrame(tabela)
    display(df)
```
